```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
```

## Load Cached Data from Local Storage

```python
# META DATA CSV 

# check needed columns
metadata_cols = pd.read_csv('data/git_data/metadata.csv', nrows=0).columns
print(metadata_cols)

# load metadata as df
meta_df = pd.read_csv('data/git_data/metadata.csv',
                      usecols=['building_id', 'site_id', 'primaryspaceusage', 'electricity', 'sqm'],
                      index_col='building_id'
                     )
```

```python
# filter building to include only Office or Education buildings

meta_df = meta_df[(meta_df['site_id'] == 'Panther') & 
    (meta_df['primaryspaceusage'].isin(['Education', 'Office'])) &
    (meta_df['electricity'] == 'Yes')]

print(meta_df.shape)
meta_df.sample(5)
```

```python
# ELECTRICITY CSV

# check the electrical buildings csv for specific columns
elec_cols = pd.read_csv('data/git_data/electricity.csv', nrows=0).columns
print(len(elec_cols))

# search for number of columns with Panther and (Education or Office)
# include the timestamp column for analysis
search_pattern = r'^timestamp$|Panther.*(?:education|office)'
matched_cols = elec_cols[elec_cols.str.contains(search_pattern, regex=True)]

print(len(matched_cols))
```

```python
# load electric consumption profiles of education and office buildings
elec_df = pd.read_csv('data/git_data/electricity.csv',
                      index_col=0,
                      usecols=matched_cols,
                      parse_dates=True,
                     )

print(elec_df.shape)
elec_df.sample(10)
```

```python
# WEATHER CSV

# filter the columns of the weather csv

w_cols = pd.read_csv('data/git_data/weather.csv', nrows=0).columns
print(w_cols)

# site_id must correspond to Panther site
weather_df = pd.read_csv('data/git_data/weather.csv', 
                         index_col=0,
                         usecols=['timestamp', 'site_id', 'airTemperature', 'dewTemperature',],
                         parse_dates=True
                        )
weather_df = weather_df[weather_df['site_id']=='Panther']
print(weather_df.shape)
weather_df.head()
```

### Normalize the Building Electrical Profiles

To enable a fair comparison across buildings of different sizes, divide the hourly electricity consumption ($\text{kWh}$) by the gross floor area ($\text{m}^2$) to calculate the hourly specific electrical intensity ($\text{kWh/m}^2$):

$$e_t = \frac{E_t}{\text{Gross Floor Area}} \quad \left[\frac{\text{kWh}}{\text{m}^2}\right]$$

Where:
- $E_t$ is the electricity consumed at hour $t$ ($\text{kWh}$)
- $e_t$ is the area-normalized electricity intensity at hour $t$ ($\text{kWh/m}^2$)

```python
# apply meta_df sqm across elec_df to normalize the buildings
meta_sqm = meta_df['sqm']
elec_normalized = elec_df.div(meta_sqm, axis=1)
elec_normalized.sample(5)
```

### Merge Weather DF with Electricity DF on time
To enable comparison of temperature data with electricity consumption across different time of the day

```python
main_df = pd.concat([elec_df, weather_df], axis=1)

# drop site_id column because all buildings are Panther
main_df.drop('site_id', inplace=True, axis=1)

print(main_df.shape)
main_df.head()
```

## Building Typology Analysis
To carry out analysis, we need to seperate buildings based on their usage and schedules. 

### Group buildings based on usage

```python
edu_cols = main_df.filter(regex=r'education').columns
office_cols = main_df.filter(regex=r'office').columns

# 1. education building
main_df['mean_edu'] = main_df[edu_cols].mean(axis=1)
main_df['median_edu'] = main_df[edu_cols].median(axis=1)
main_df['Q25_edu'] = main_df[edu_cols].quantile(0.25, axis=1)
main_df['Q75_edu'] = main_df[edu_cols].quantile(0.75, axis=1)

# 2. office building
main_df['mean_office'] = main_df[office_cols].mean(axis=1)
main_df['median_office'] = main_df[office_cols].median(axis=1)
main_df['Q25_office'] = main_df[office_cols].quantile(0.25, axis=1)
main_df['Q75_office'] = main_df[office_cols].quantile(0.75, axis=1)

feature_cols = [
    'airTemperature', 'dewTemperature', 'mean_edu',
    'median_edu', 'Q25_edu', 'Q75_edu', 'mean_office',
    'median_office', 'Q25_office', 'Q75_office'
]

analysis_df = main_df[feature_cols].copy()
analysis_df.sample(10)
```

### Group Buildings based on Schedules

```python
analysis_df['year'] = analysis_df.index.year
analysis_df['hour'] = analysis_df.index.hour
analysis_df['weekday'] = analysis_df.index.weekday
analysis_df['day_name'] = analysis_df.index.day_name()
analysis_df['is_weekend'] = analysis_df['weekday'] >= 5

# infer periods where building is mostly likely occupied
analysis_df['is_occupied'] = (~analysis_df['is_weekend']) & (analysis_df['hour'].between(8, 17))

analysis_df.sample(10)
```

```python
# add meteorological seasons
season_map = {
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Fall', 10: 'Fall', 11: 'Fall'
}

analysis_df['season'] = analysis_df.index.month.map(season_map)

analysis_df.head()
```

## Plot Graphs
Year is seperated into 2017 and 2018 for easy comparison across the both year

```python
fig, ((edu_16, edu_17), (office_16, office_17)) = plt.subplots(2, 2, sharex=True, sharey=True, figsize=(8, 6))

# Boolean masks
occ = analysis_df['is_occupied']
unocc = ~analysis_df["is_occupied"]

# 2017 vs 2018
df_2016 = analysis_df[analysis_df['year']==2016].copy()
df_2017 = analysis_df[analysis_df['year']==2017].copy()


# Education Building 2016
edu_16.scatter(
    df_2016.loc[unocc, 'airTemperature'],
    df_2016.loc[unocc, 'mean_edu'],
    color='#7570B3',
    alpha=0.3,
    s=15,
    label='Unoccupied',
)
edu_16.scatter(
    df_2016.loc[occ, 'airTemperature'],
    df_2016.loc[occ, 'mean_edu'],
    color='#E66101',
    alpha=0.4,
    s=15,
    label='Occupied',
)
edu_16.set_title('Education 2016')

# Education Building 2017
edu_17.scatter(
    df_2017.loc[unocc, 'airTemperature'],
    df_2017.loc[unocc, 'mean_edu'],
    color='#7570B3',
    alpha=0.3,
    s=15,
)
edu_17.scatter(
    df_2017.loc[occ, 'airTemperature'],
    df_2017.loc[occ, 'mean_edu'],
    color='#E66101',
    alpha=0.4,
    s=15,
)
edu_17.set_title('Education 2017')



# Office Building 2016
office_16.scatter(
    df_2016.loc[unocc, 'airTemperature'],
    df_2016.loc[unocc, 'mean_office'],
    color='#7570B3',
    alpha=0.3,
    s=15,
)
office_16.scatter(
    df_2016.loc[occ, 'airTemperature'],
    df_2016.loc[occ, 'mean_office'],
    color='#E66101',
    alpha=0.4,
    s=15,
)
office_16.set_title('Office 2016')

# Office Building 2017
office_17.scatter(
    df_2017.loc[unocc, 'airTemperature'],
    df_2017.loc[unocc, 'mean_office'],
    color='#7570B3',
    alpha=0.3,
    s=15,
)
office_17.scatter(
    df_2017.loc[occ, 'airTemperature'],
    df_2017.loc[occ, 'mean_office'],
    color='#E66101',
    alpha=0.4,
    s=15,
)
office_17.set_title('Office 2017')



fig.suptitle('Does Temperature influence electricity consumption of education buildings or office buildings?',
             fontweight='bold',
             style='italic',
             x=0.5,
             y=1.03,             
            )
fig.supxlabel('Outdoor Temperature (°C)')
fig.supylabel('Energy (kWh/m²)')
fig.legend(loc='upper center', 
           ncol=2, 
           bbox_to_anchor=(0.5, 1),
           frameon=False
          )
plt.tight_layout()
plt.show()
```
