```python
import pandas as pd
import requests
import zipfile
import io

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00357/occupancy_data.zip'
response = requests.get(url)

byte_data = io.BytesIO(response.content) # mimicks file placement in hard drive (virtual file)

zip_folder = zipfile.ZipFile(byte_data)

print(zip_folder.namelist())
```

```python
df = pd.read_csv(zip_folder.open('datatraining.txt'), sep=',')

print(len(df))
df.columns = [x.lower().strip() for x in df.columns] #clean column names
df.head()
```

## <u>***Correlation of Occupancy with Light, Temperature and CO2***</u>

```python
import scipy.stats as stats
```

```python
corr_l, pval_l = stats.pearsonr(df['light'], df['occupancy']) # correlation between light and occupancy
corr_t, pval_t = stats.pearsonr(df['temperature'], df['occupancy']) # correlation between temperature and occupancy
corr_c, pval_c = stats.pearsonr(df['co2'], df['occupancy']) # correlation between co2 and occupancy

print(f"Correlation between light and occupancy: {corr_l:.2f}")
print(f"Correlation between temperature and occupancy: {corr_t:.2f}")
print(f"Correlation between CO2 and occupancy: {corr_c:.2f}")
```

##### The correlation analysis reveals that: 

1. **lighting state** is the most reliable indicator of room occupancy ***(r = 0.91)***. This aligns with expected occupant behavior, as lighting is an immediate, manual response to entering a space. 

2. **CO2 concentration** serves as a strong secondary metric ***(r = 0.71)***, accurately reflecting human metabolic load, though it is subject to the room's ventilation rate. 

3. **Temperature** showed the weakest relationship to immediate occupancy ***(r = 0.54)***, likely due to the thermal mass of the building and HVAC system lag.
