```python
import csv

%precision 2 #display not more than 2dp! Doesn't affect calculation

with open('IAQ_dummy.csv') as csvfile: #using 'with' automatically closes our file
    iaq_params = list(csv.DictReader(csvfile)) #list() ensures we consume the iterator, without it we will only be returning the location of the iterator object

iaq_params[:3] #print first 3 dict rows
```

**Mean, Max and Min Temperature for Lecture Hall A**

```python
mean_temp = (sum(float(iaq['Temperature_C']) for iaq in iaq_params) / len(iaq_params)) #average temp
max_temp = max(float(iaq['Temperature_C']) for iaq in iaq_params) #max temp
min_temp = min(float(iaq['Temperature_C']) for iaq in iaq_params) #min temp

print(f'The mean temperature is {round(mean_temp, 2)}')
print(f'The maximum temperature is {round(max_temp, 2)}')
print(f'The minimum temperature is {round(min_temp, 2)}')
```

**When CO2 concentration exceeds the commonly referenced 1000ppm value**

```python
high_co2 = [iaq['CO2_ppm'] for iaq in iaq_params if float(iaq['CO2_ppm']) > 1000]
count = len(high_co2)
print(set(high_co2)) #ensure there is no duplicate
print(f'CO2 exceeded 1000ppm {count} number of times')
```

**Flag the time with High CO2 concentration**

```python
high_co2_pairs = []
for co2 in high_co2:
  for iaq in iaq_params:
    if iaq['CO2_ppm'] == co2:
      high_co2_pairs.append((iaq['Timestamp'], iaq['CO2_ppm'])) #append tuple ('date-time-stamp', 'high_co2')
high_co2_pairs.sort(key=lambda x: x[1], reverse=True)
high_co2_pairs


```
